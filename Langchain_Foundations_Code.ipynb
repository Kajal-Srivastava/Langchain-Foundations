```python
from langchain.chat_models import init_chat_model
model = init_chat_model(model="gpt-5-nano")
```

```python
response = model.invoke("What is the capital of the Moon?")
response
```

```python
#to print the metadata details
from pprint import pprint
pprint(response.response_metadata)
```

```python
#Using temperature parameter
model = init_chat_model(
model = "gpt-5-nano",
temperature = 1.0
)
response = model.invoke("what is the capital of moon?")
print(response.content)
```

```python
#initializing an agent
from langchain.agents import create_agent
agent = create_agent(model=model)
```

```python
agent = create_agent("gpt-5-nano")
```

```python
from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages":[HumanMessage(content="What is the capital of the moon?")]}
    
)
response
```

```python
from pprint import pprint
pprint(response)
```

```python
#Streaming output to reduce the output duration or show messages in a streamed way
for token, metadata in agent.stream (
     {"messages":[HumanMessage(content="Tell me about Luna City,the capital of The Moon?")]},
    stream_mode = "messages"
):
    if token.content:
        print(token.content,end="",flush=True)
    
```
